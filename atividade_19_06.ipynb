```python
#QUESTÃO 02 – Soma dos Números (2 pontos)
#Solicite ao usuário um número inteiro positivo.
#utilizando um laço de repetição, calcule e exiba a soma de todos os números de 01 até o número informado

while True:
    numero  = int(input("Digite um numero inteiro positivo: "))
    if numero > 0:
        break
    else:
        print("Por favor, digite um número inteiro positivo.")

soma = 0
sum = ""
for i in range (1,numero + 1):
  soma = soma + i
  sum += str(i)
  if i < numero:
    sum += " + "
print(f'Resultado: {soma}')
print(f'Pois: {sum} = {soma}')
```

```python
#QUESTÃO 04 – Fatorial (4 pontos)
#---------------------------------------------------------

#Solicite ao usuário um número inteiro positivo e calcule
#seu fatorial utilizando um laço de repetição

numb = int  (input("Digite um numero inteiro positivo: "))
while numb < 0:
  numb = int  (input("digite um numero inteiro positivo: "))
fatorial = 1
for i in range (1,numb+1):
  fatorial = fatorial * i
print(f'Fatorial = {fatorial}')
print('Pois:')
print(f'{numb} = ',end='')
for i in range (numb,1,-1):
  print(f'{i} × ',end='')
print('1')
```

```python
#QUESTÃO 05 – Caixa Eletrônico (5 pontos)
#---------------------------------------------------------

#Um caixa eletrônico possui notas de:

#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

#Solicite ao usuário o valor do saque e informe a menor
#quantidade possível de notas para realizar o pagamento.

brs = ("B Rosa ")
print("-=" * 30)
print(f"-=" *12, f"{brs}", "=-" * 12)
valor = int(input("Digite o Valor do Saque:  R$: "))
total = valor
ced = 100
totalced = 0
while True:
  if total >= ced:
    total -= ced
    totalced += 1
  else:
    if totalced > 0:
      print(f'Total de {totalced} cédulas de R$: {ced}')
    if ced == 100:
      ced = 50
    if ced == 50:
     ced = 20
    elif ced == 20:
      ced = 10
    elif ced == 10:
      ced = 1
    totalced = 0
    if total == 0:
      break
print('-=' * 30)
```
